```python
import torch
import gc
from PIL import Image
import requests
import numpy as np
import matplotlib.pyplot as plt
import torch.nn.functional as F

def flush_memory():
    """Utility to clear GPU memory between exercises to avoid OOM."""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.ipc_collect()
    print("Memory flushed.")
```

## 2. HuggingFace Pipeline API

The `pipeline` API is the fastest way to use a model. Let's classify an image of a bus.

```python
from transformers import pipeline

# Exercise 1: Zero-shot Image Classification
classifier = pipeline("image-classification", model="google/vit-base-patch16-224")
url = "https://ultralytics.com/images/bus.jpg"
result = classifier(url)

for r in result:
    print(f"{r['label']}: {r['score']:.4f}")
```

```python
# clean the memory for the next models
del classifier
flush_memory()
```

## 2. Vision Embeddings (CLIP)

Embeddings map images into a semantic space. CLIP connects text to images.

```python
from transformers import CLIPModel, CLIPProcessor, AutoModel, AutoImageProcessor


# Exercise 3: CLIP Zero-shot Classification
clip_id = "openai/clip-vit-base-patch32"
clip_model = CLIPModel.from_pretrained(clip_id).to("cuda" if torch.cuda.is_available() else "cpu")
clip_proc = CLIPProcessor.from_pretrained(clip_id)

image = Image.open(requests.get("https://ultralytics.com/images/bus.jpg", stream=True).raw)
labels = ["a photo of a bus", "a photo of a city scene", "a photo of a desert"]
inputs = clip_proc(text=labels, images=image, return_tensors="pt", padding=True).to(clip_model.device)

with torch.no_grad():
    outputs = clip_model(**inputs)
    probs = outputs.logits_per_image.softmax(dim=1)

for label, prob in zip(labels, probs[0]):
    print(f"{label}: {prob.item():.4f}")

del clip_model, clip_proc
flush_memory()
```

## 3. Vision Language Models (Qwen3.5-0.8B)

The light-weight Qwen3.5-0.8B is an excelente VLM for on-device reasoning and OCR.

```python
from transformers import AutoProcessor, AutoModelForImageTextToText

processor = AutoProcessor.from_pretrained("Qwen/Qwen3.5-0.8B")
model = AutoModelForImageTextToText.from_pretrained("Qwen/Qwen3.5-0.8B")

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": "https://ultralytics.com/images/bus.jpg"},
            {"type": "text", "text": "Describe what color the vehicles in this image are."},
        ],
    }
]

inputs = processor.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=40)
print(processor.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```

```python
del model, processor
flush_memory()
```

## 4. Spatial Features & Patch Relationships (TIPSv2)

Google's **TIPSv2** (Text-Image Pre-training with Spatial awareness) family is designed to produce spatially rich image features. Unlike standard CLIP, which primarily optimizes for global image-text alignment, TIPSv2 focuses on dense patch-text alignment.

This makes TIPSv2 excellent for:
*   **Zero-shot Image Classification** (often outperforming CLIP).
*   **Dense Prediction Tasks** (depth estimation, segmentation).
*   **Semantic Patch Relationships** (similar to DINOv3, but aligned with text).

In this exercise, we will visualize how TIPSv2 groups patches semantically using PCA.

```python
# Exercise 4: Visualizing Patch Relationships with PCA
# Similar to DINOv3, we can look at the patch embeddings to see semantic segments.
from transformers import AutoModel, AutoImageProcessor
from sklearn.decomposition import PCA
import torch.nn.functional as F

model_id = "google/tipsv2-b14"
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load model and processor
tips_model = AutoModel.from_pretrained(model_id, trust_remote_code=True).to(device).eval()

# Process image
image = Image.open(requests.get("https://ultralytics.com/images/bus.jpg", stream=True).raw)
img_size = 448 # TIPSv2 standard size
pixel_values = F.interpolate(torch.tensor(np.array(image.convert("RGB"))).permute(2, 0, 1).unsqueeze(0).float() / 255.0, 
                             size=(img_size, img_size), mode='bilinear', align_corners=False).to(device)

# Extract patch tokens
with torch.no_grad():
    outputs = tips_model.encode_image(pixel_values)
    # patch_tokens shape: (1, 1024, 768) for 448x448 img with 14x14 patches
    patches = outputs.patch_tokens[0].cpu().numpy() 

# Apply PCA to reduce 768-dim to 3-dim (RGB)
pca = PCA(n_components=3)
pca_features = pca.fit_transform(patches)

# Rescale to [0, 1] for visualization
pca_features = (pca_features - pca_features.min(0)) / (pca_features.max(0) - pca_features.min(0))

# Reshape back to grid (32x32 patches for 448/14)
h, w = img_size // 14, img_size // 14
pca_img = pca_features.reshape(h, w, 3)

# Plot
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(pca_img)
plt.title("TIPSv2 Patch PCA (Semantic Segments)")
plt.axis("off")
plt.show()

del tips_model
flush_memory()
```
